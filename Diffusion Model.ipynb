```python
from diffusers import StableDiffusionPipeline
import matplotlib.pyplot as plt
```

```python
import torch
```

```python
import os
os.environ['CURL_CA_BUNDLE'] = ''
os.environ['REQUESTS_CA_BUNDLE'] = ''
```

```python
model= "dreamlike-art/dreamlike-diffusion-1.0"

pipe= StableDiffusionPipeline.from_pretrained(model, torch_type= torch.float16, resume_download=True)
pipe= pipe.to("cpu")
```

```python
prompt= "Lord Shiva with his terrific form of Shiv Tandav"

image= pipe(prompt).images[0]
```

```python
print("[PROMPT] :", prompt)
plt.imshow(image)
plt.style.use('dark_background')
plt.axis("off")
```
