```python
from transformers import pipeline
clf= pipeline('sentiment-analysis')
```

```python
clf("Huggingface is a great open source library API to integrate with programs and to leverage state of the art results")
```

```python
results = clf(["This Hugginface API is awesome and we can integrate pre-trained models easily!", 
              "But Huggingface is tough for beginners and is not beginner friendly"])

for result in results:
    print(f" label: {result['label']},  with score: {round(result['score'], 3)}")
```

```python
clf(["The pizza is not that great but the crust is ok", "The decision was tough but it changed my life"])
```

```python
from transformers import AutoTokenizer, TFAutoModelForSequenceClassification
```

```python
model_name= "nlptown/bert-base-multilingual-uncased-sentiment"
model= TFAutoModelForSequenceClassification.from_pretrained(model_name, from_pt= True)
tokenizer= AutoTokenizer.from_pretrained(model_name)
clf= pipeline('sentiment-analysis', model=model, tokenizer=tokenizer)
```

```python
clf("We hope you will not hate it")
```

```python
model_name= "distilbert/distilbert-base-uncased-finetuned-sst-2-english"
model= TFAutoModelForSequenceClassification.from_pretrained(model_name, from_pt= True)
tokenizer= AutoTokenizer.from_pretrained(model_name)
clf= pipeline('sentiment-analysis', model=model, tokenizer=tokenizer)
```
